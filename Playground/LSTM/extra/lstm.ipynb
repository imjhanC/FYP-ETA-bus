```python
import pandas as pd
import numpy as np
import time
from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.metrics import (mean_absolute_error, mean_squared_error, r2_score,
                            mean_absolute_percentage_error, explained_variance_score,
                            max_error, median_absolute_error)
from sklearn.preprocessing import StandardScaler, LabelEncoder
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense, Dropout
from tensorflow.keras.callbacks import EarlyStopping
import optuna
import joblib
import matplotlib.pyplot as plt

# Load the data
data = pd.read_csv('preprocessed.csv')

# Define features and target
features = ['current_stop_name', 'next_stop_name', 'day_of_week', 'is_holiday', 
            'is_peak_hour', 'weather_condition', 'passenger_count', 'current_speed', 
            'distance_to_next_stop', 'current_lat', 'current_lon']
target = 'eta_minutes'

X = data[features].copy()
y = data[target].values

# Preprocess categorical features
categorical_cols = ['current_stop_name', 'next_stop_name', 'day_of_week', 'weather_condition']
label_encoders = {}
for col in categorical_cols:
    le = LabelEncoder()
    X[col] = le.fit_transform(X[col])
    label_encoders[col] = le
    
# Convert boolean columns to int
X['is_holiday'] = X['is_holiday'].astype(int)
X['is_peak_hour'] = X['is_peak_hour'].astype(int)

# Scale features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Reshape input for LSTM [samples, time steps, features]
X_reshaped = X_scaled.reshape(X_scaled.shape[0], 1, X_scaled.shape[1])

# Split data
X_train, X_test, y_train, y_test = train_test_split(X_reshaped, y, test_size=0.2, random_state=42)

# Enhanced evaluation function with comprehensive metrics
def evaluate_model(model, X_train, y_train, X_test, y_test, training_time=None):
    start_time = time.time()
    
    # Prediction time measurement
    pred_start_time = time.time()
    y_pred = model.predict(X_test)
    prediction_time = time.time() - pred_start_time
    
    # Train predictions for overfitting analysis
    y_train_pred = model.predict(X_train)
    
    # Flatten predictions if needed
    if len(y_pred.shape) > 1:
        y_pred = y_pred.flatten()
        y_train_pred = y_train_pred.flatten()
    
    # 1. Primary Regression Metrics
    mae = mean_absolute_error(y_test, y_pred)
    mse = mean_squared_error(y_test, y_pred)
    rmse = np.sqrt(mse)
    r2 = r2_score(y_test, y_pred)
    
    # Only calculate MAPE if there are no zeros in y_test
    if not np.any(y_test == 0):
        mape = mean_absolute_percentage_error(y_test, y_pred) * 100
    else:
        # Handle zeros in y_test by excluding them from MAPE calculation
        non_zero_indices = y_test != 0
        if np.any(non_zero_indices):
            mape = mean_absolute_percentage_error(y_test[non_zero_indices], y_pred[non_zero_indices]) * 100
        else:
            mape = float('nan')
    
    explained_var = explained_variance_score(y_test, y_pred)
    max_err = max_error(y_test, y_pred)
    median_abs_err = median_absolute_error(y_test, y_pred)
    
    # 3. Error Distribution Metrics
    abs_errors = np.abs(y_test - y_pred)
    p90_error = np.percentile(abs_errors, 90)
    p95_error = np.percentile(abs_errors, 95)
    p99_error = np.percentile(abs_errors, 99)
    
    # 4. Overfitting Metrics
    train_mse = mean_squared_error(y_train, y_train_pred)
    train_rmse = np.sqrt(train_mse)
    overfitting_ratio = train_rmse / rmse if rmse > 0 else float('inf')
    
    # 5. Residual Analysis Metrics
    residuals = y_test - y_pred
    residuals_mean = residuals.mean()
    residuals_std = residuals.std()
    
    # 6. Performance metrics
    evaluation_time = time.time() - start_time
    
    # Print metrics
    print("\n" + "="*50)
    print("MODEL EVALUATION METRICS")
    print("="*50)
    
    print("\n1. PRIMARY REGRESSION METRICS:")
    print(f"MAE: {mae:.4f}")
    print(f"MSE: {mse:.4f}")
    print(f"RMSE: {rmse:.4f}")
    print(f"MAPE: {mape:.4f}%" if not np.isnan(mape) else "MAPE: N/A (zero values in target)")
    print(f"R² Score: {r2:.4f}")
    print(f"Explained Variance: {explained_var:.4f}")
    print(f"Max Error: {max_err:.4f}")
    print(f"Median Absolute Error: {median_abs_err:.4f}")
    
    print("\n3. ERROR DISTRIBUTION METRICS:")
    print(f"90th Percentile of Absolute Errors (P90): {p90_error:.4f}")
    print(f"95th Percentile of Absolute Errors (P95): {p95_error:.4f}")
    print(f"99th Percentile of Absolute Errors (P99): {p99_error:.4f}")
    
    print("\n4. OVERFITTING METRICS:")
    print(f"Train RMSE: {train_rmse:.4f}")
    print(f"Test RMSE: {rmse:.4f}")
    print(f"Overfitting Ratio (Train RMSE / Test RMSE): {overfitting_ratio:.4f}")
    print(f"Interpretation: {'Potential underfitting' if overfitting_ratio > 1.1 else 'Potential overfitting' if overfitting_ratio < 0.9 else 'Good generalization'}")
    
    print("\n5. RESIDUAL ANALYSIS METRICS:")
    print(f"Mean of Residuals: {residuals_mean:.4f}")
    print(f"Standard Deviation of Residuals: {residuals_std:.4f}")
    
    print("\n6. PERFORMANCE METRICS:")
    print(f"Prediction Time (s): {prediction_time:.4f}")
    if training_time is not None:
        print(f"Training Time (s): {training_time:.4f}")
    print(f"Evaluation Time (s): {evaluation_time:.4f}")
    
    # Create a dictionary with all metrics for comparison
    metrics = {
        'MAE': mae,
        'MSE': mse,
        'RMSE': rmse,
        'MAPE': mape if not np.isnan(mape) else None,
        'R2': r2,
        'Explained_Variance': explained_var,
        'Max_Error': max_err,
        'Median_Absolute_Error': median_abs_err,
        'P90_Error': p90_error,
        'P95_Error': p95_error,
        'P99_Error': p99_error,
        'Train_RMSE': train_rmse,
        'Overfitting_Ratio': overfitting_ratio,
        'Residuals_Mean': residuals_mean,
        'Residuals_Std': residuals_std,
        'Prediction_Time': prediction_time,
        'Training_Time': training_time,
        'Evaluation_Time': evaluation_time
    }
    
    return metrics

# Function to plot residuals and error distributions
def plot_model_diagnostics(y_test, y_pred, model_name):
    residuals = y_test - y_pred
    abs_errors = np.abs(residuals)
    
    fig, axes = plt.subplots(2, 2, figsize=(16, 12))
    
    # Predicted vs Actual
    axes[0, 0].scatter(y_test, y_pred, alpha=0.5)
    axes[0, 0].plot([min(y_test), max(y_test)], [min(y_test), max(y_test)], 'r--')
    axes[0, 0].set_xlabel('Actual Values')
    axes[0, 0].set_ylabel('Predicted Values')
    axes[0, 0].set_title('Predicted vs Actual Values')
    
    # Residuals plot
    axes[0, 1].scatter(y_pred, residuals, alpha=0.5)
    axes[0, 1].axhline(y=0, color='r', linestyle='--')
    axes[0, 1].set_xlabel('Predicted Values')
    axes[0, 1].set_ylabel('Residuals')
    axes[0, 1].set_title('Residuals vs Predicted Values')
    
    # Histogram of residuals
    axes[1, 0].hist(residuals, bins=30, alpha=0.7, edgecolor='black')
    axes[1, 0].axvline(x=0, color='r', linestyle='--')
    axes[1, 0].set_xlabel('Residuals')
    axes[1, 0].set_ylabel('Frequency')
    axes[1, 0].set_title('Distribution of Residuals')
    
    # Error distribution (absolute errors)
    axes[1, 1].hist(abs_errors, bins=30, alpha=0.7, edgecolor='black', cumulative=True, density=True)
    axes[1, 1].axvline(x=np.percentile(abs_errors, 90), color='r', linestyle='--', 
                     label=f'P90: {np.percentile(abs_errors, 90):.2f}')
    axes[1, 1].axvline(x=np.percentile(abs_errors, 95), color='g', linestyle='--', 
                     label=f'P95: {np.percentile(abs_errors, 95):.2f}')
    axes[1, 1].axvline(x=np.percentile(abs_errors, 99), color='b', linestyle='--', 
                     label=f'P99: {np.percentile(abs_errors, 99):.2f}')
    axes[1, 1].set_xlabel('Absolute Error')
    axes[1, 1].set_ylabel('Cumulative Density')
    axes[1, 1].set_title('Cumulative Distribution of Absolute Errors')
    axes[1, 1].legend()
    
    plt.tight_layout()
    plt.savefig(f'{model_name}_diagnostics.png')
    print(f"Model diagnostics saved as '{model_name}_diagnostics.png'")
    return fig

# Train baseline LSTM model
def create_baseline_model(input_shape):
    model = Sequential([
        LSTM(50, return_sequences=True, input_shape=input_shape),
        Dropout(0.2),
        LSTM(50),
        Dropout(0.2),
        Dense(1)
    ])
    
    model.compile(optimizer='adam', loss='mse', metrics=['mae'])
    return model

print("Training baseline LSTM model...")
input_shape = (X_train.shape[1], X_train.shape[2])
baseline_model = create_baseline_model(input_shape)

# Early stopping to prevent overfitting
early_stopping = EarlyStopping(monitor='val_loss', patience=10, restore_best_weights=True)

# Train baseline model with timing
baseline_start_time = time.time()
baseline_history = baseline_model.fit(
    X_train, y_train,
    validation_split=0.2,
    epochs=100,
    batch_size=32,
    callbacks=[early_stopping],
    verbose=1
)
baseline_training_time = time.time() - baseline_start_time

print("\nBaseline Model Performance:")
baseline_metrics = evaluate_model(baseline_model, X_train, y_train, X_test, y_test, baseline_training_time)

# Plot baseline model diagnostics
y_baseline_pred = baseline_model.predict(X_test).flatten()
baseline_diagnostics = plot_model_diagnostics(y_test, y_baseline_pred, 'baseline_lstm')

# Optuna optimization for LSTM
def objective(trial):
    # Define hyperparameters to tune
    lstm_units_1 = trial.suggest_int('lstm_units_1', 32, 128)
    lstm_units_2 = trial.suggest_int('lstm_units_2', 16, 64)
    dropout_rate_1 = trial.suggest_float('dropout_rate_1', 0.1, 0.5)
    dropout_rate_2 = trial.suggest_float('dropout_rate_2', 0.1, 0.5)
    learning_rate = trial.suggest_float('learning_rate', 1e-4, 1e-2, log=True)
    batch_size = trial.suggest_categorical('batch_size', [16, 32, 64])
    
    # Create model with trial hyperparameters
    model = Sequential([
        LSTM(lstm_units_1, return_sequences=True, input_shape=input_shape),
        Dropout(dropout_rate_1),
        LSTM(lstm_units_2),
        Dropout(dropout_rate_2),
        Dense(1)
    ])
    
    # Compile model
    optimizer = tf.keras.optimizers.Adam(learning_rate=learning_rate)
    model.compile(optimizer=optimizer, loss='mse', metrics=['mae'])
    
    # Train model
    early_stopping = EarlyStopping(monitor='val_loss', patience=5, restore_best_weights=True)
    history = model.fit(
        X_train, y_train,
        validation_split=0.2,
        epochs=50,
        batch_size=batch_size,
        callbacks=[early_stopping],
        verbose=0
    )
    
    # Evaluate on validation data
    val_rmse = np.sqrt(min(history.history['val_loss']))
    
    return val_rmse

print("\nStarting Optuna optimization for LSTM model...")
optuna_start_time = time.time()
study = optuna.create_study(direction='minimize')
study.optimize(objective, n_trials=30)
optuna_time = time.time() - optuna_start_time

best_params = study.best_params
print(f"Best parameters: {best_params}")
print(f"Optuna optimization time: {optuna_time:.2f} seconds")

# Train optimized model
print("\nTraining optimized LSTM model with best parameters...")
optimized_model = Sequential([
    LSTM(best_params['lstm_units_1'], return_sequences=True, input_shape=input_shape),
    Dropout(best_params['dropout_rate_1']),
    LSTM(best_params['lstm_units_2']),
    Dropout(best_params['dropout_rate_2']),
    Dense(1)
])

optimizer = tf.keras.optimizers.Adam(learning_rate=best_params['learning_rate'])
optimized_model.compile(optimizer=optimizer, loss='mse', metrics=['mae'])

# Train with early stopping and timing
optimized_start_time = time.time()
early_stopping = EarlyStopping(monitor='val_loss', patience=15, restore_best_weights=True)
optimized_history = optimized_model.fit(
    X_train, y_train,
    validation_split=0.2,
    epochs=150,
    batch_size=best_params['batch_size'],
    callbacks=[early_stopping],
    verbose=1
)
optimized_training_time = time.time() - optimized_start_time

print("\nOptimized Model Performance:")
optimized_metrics = evaluate_model(optimized_model, X_train, y_train, X_test, y_test, optimized_training_time)

# Plot optimized model diagnostics
y_optimized_pred = optimized_model.predict(X_test).flatten()
optimized_diagnostics = plot_model_diagnostics(y_test, y_optimized_pred, 'optimized_lstm')

# Compare baseline and optimized models
print("\nPERFORMANCE COMPARISON:")
print("="*50)
comparison_metrics = [
    'RMSE', 'MAE', 'R2', 'MAPE', 'P90_Error', 'P95_Error', 
    'Overfitting_Ratio', 'Training_Time'
]

print(f"{'Metric':<20}{'Baseline':<15}{'Optimized':<15}{'Improvement':<15}{'Improvement %':<15}")
print("-"*70)

for metric in comparison_metrics:
    if metric in baseline_metrics and metric in optimized_metrics:
        baseline_value = baseline_metrics[metric]
        optimized_value = optimized_metrics[metric]
        
        if baseline_value is not None and optimized_value is not None:
            if metric == 'R2' or metric == 'Explained_Variance':  # Higher is better
                improvement = optimized_value - baseline_value
                improvement_pct = (improvement / abs(baseline_value)) * 100 if baseline_value != 0 else float('inf')
            elif metric == 'Overfitting_Ratio':  # Closer to 1 is better
                improvement = abs(baseline_value - 1) - abs(optimized_value - 1)
                improvement_pct = (improvement / abs(baseline_value - 1)) * 100 if baseline_value != 1 else float('inf')
            else:  # Lower is better for other metrics
                improvement = baseline_value - optimized_value
                improvement_pct = (improvement / baseline_value) * 100 if baseline_value != 0 else float('inf')
                
            print(f"{metric:<20}{baseline_value:<15.4f}{optimized_value:<15.4f}{improvement:<15.4f}{improvement_pct:<15.2f}%")
        else:
            print(f"{metric:<20}{str(baseline_value):<15}{str(optimized_value):<15}{'N/A':<15}{'N/A':<15}")

# Save the optimized model
model_filename = 'optimized_lstm_eta_predictor.h5'
optimized_model.save(model_filename)
print(f"\nOptimized model saved as {model_filename}")

# Save preprocessors for later use
joblib.dump(scaler, 'scaler.pkl')
joblib.dump(label_encoders, 'label_encoders.pkl')
print("Preprocessing objects saved")

# Visualize training history
plt.figure(figsize=(12, 10))

plt.subplot(2, 2, 1)
plt.plot(optimized_history.history['loss'], label='Training Loss')
plt.plot(optimized_history.history['val_loss'], label='Validation Loss')
plt.title('Loss Over Time')
plt.xlabel('Epochs')
plt.ylabel('Loss (MSE)')
plt.legend()

plt.subplot(2, 2, 2)
plt.plot(optimized_history.history['mae'], label='Training MAE')
plt.plot(optimized_history.history['val_mae'], label='Validation MAE')
plt.title('Mean Absolute Error Over Time')
plt.xlabel('Epochs')
plt.ylabel('MAE')
plt.legend()

plt.subplot(2, 2, 3)
plt.plot(baseline_history.history['loss'], label='Baseline Loss')
plt.plot(optimized_history.history['loss'], label='Optimized Loss')
plt.title('Training Loss Comparison')
plt.xlabel('Epochs')
plt.ylabel('Loss (MSE)')
plt.legend()

plt.subplot(2, 2, 4)
plt.plot(baseline_history.history['val_loss'], label='Baseline Val. Loss')
plt.plot(optimized_history.history['val_loss'], label='Optimized Val. Loss')
plt.title('Validation Loss Comparison')
plt.xlabel('Epochs')
plt.ylabel('Loss (MSE)')
plt.legend()

plt.tight_layout()
plt.savefig('lstm_training_history_comparison.png')
print("Training history visualization saved as 'lstm_training_history_comparison.png'")

# Create a learning curve visualization to show model performance with training size
def plot_learning_curves(X_train, y_train, X_test, y_test):
    training_sizes = np.linspace(0.1, 1.0, 10)
    train_errors = []
    test_errors = []
    
    for size in training_sizes:
        # Use a subset of training data
        subset_size = int(len(X_train) * size)
        X_subset = X_train[:subset_size]
        y_subset = y_train[:subset_size]
        
        # Create and train a model with the same architecture
        model = create_baseline_model(input_shape)
        model.fit(X_subset, y_subset, epochs=30, batch_size=32, verbose=0)
        
        # Calculate errors
        train_pred = model.predict(X_subset).flatten()
        test_pred = model.predict(X_test).flatten()
        
        train_rmse = np.sqrt(mean_squared_error(y_subset, train_pred))
        test_rmse = np.sqrt(mean_squared_error(y_test, test_pred))
        
        train_errors.append(train_rmse)
        test_errors.append(test_rmse)
    
    plt.figure(figsize=(10, 6))
    plt.plot(training_sizes * 100, train_errors, 'o-', color='r', label='Training error')
    plt.plot(training_sizes * 100, test_errors, 'o-', color='g', label='Testing error')
    plt.title('Learning Curves (RMSE vs Training Size)')
    plt.xlabel('Training Size (%)')
    plt.ylabel('RMSE')
    plt.legend()
    plt.grid(True)
    plt.savefig('learning_curves.png')
    print("Learning curves saved as 'learning_curves.png'")

print("\nGenerating learning curves (this may take a few minutes)...")
plot_learning_curves(X_train, y_train, X_test, y_test)

# Final summary
print("\n" + "="*50)
print("FINAL MODEL SUMMARY")
print("="*50)
print(f"Baseline RMSE: {baseline_metrics['RMSE']:.4f}")
print(f"Optimized RMSE: {optimized_metrics['RMSE']:.4f}")
print(f"Improvement: {(baseline_metrics['RMSE'] - optimized_metrics['RMSE']):.4f} " +
      f"({((baseline_metrics['RMSE'] - optimized_metrics['RMSE'])/baseline_metrics['RMSE']*100):.2f}%)")
print(f"Optuna optimization time: {optuna_time:.2f} seconds")
print(f"Total model development time: {baseline_training_time + optuna_time + optimized_training_time:.2f} seconds")
print("\nAll evaluation metrics, diagnostics plots, and model files have been saved.")
```
