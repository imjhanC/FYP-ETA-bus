```python
import pandas as pd
import joblib

# Load your trained model
model = joblib.load('optimized_lightgbm_eta_predictor.pkl')

# Create a dictionary with your custom feature values
custom_features = {
    'current_stop_name': 0,          # Stop ID (integer)
    'next_stop_name': 3,             # Next stop ID (integer)
    'day_of_week': 0,                # 0-6 (Monday-Sunday)
    'is_holiday': False,                 # 0=False, 1=True
    'is_peak_hour': True,               # 0=False, 1=True
    'weather_condition': 4,          # Numerical code (0-4)
    'passenger_count': 100,           # Integer count
    'current_speed': 15.1,           # km/h (float)
    'distance_to_next_stop': 2.6,   # km (float)
    'current_lat': 3.03917,          # Latitude (float)
    'current_lon': 101.79858         # Longitude (float)
}

# Convert to DataFrame (single row)
input_df = pd.DataFrame([custom_features])

# Make prediction
predicted_eta = model.predict(input_df)[0]
print(f"Predicted ETA: {predicted_eta:.2f} minutes")
```
